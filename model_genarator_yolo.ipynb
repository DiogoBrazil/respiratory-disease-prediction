```python
from ultralytics import YOLO
```

```python
# Load a model
model = YOLO("yolov8n-cls.pt")  # load a pretrained model (recommended for training)
```

```python
# Train the model
results = model.train(data="Departamento médico", epochs=30, imgsz=256)
```

```python
metrics = model.metrics
```
